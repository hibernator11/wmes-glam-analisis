```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('GLAM-WMES-Actividades.csv')
```

```python
df
```

```python
df.dtypes
```

```python
df['Fecha'] = pd.to_datetime(df['Fecha'], format='mixed', errors='coerce')
```

```python
df
```

```python
df['Ano'] = df['Fecha'].apply(lambda x: x.strftime('%Y'))
```

```python
df.groupby(['Ano'])['Ano'].count()
```

```python
df.groupby(['Ano'])['Ano'].count().plot.bar()

plt.xlabel('Año')
plt.ylabel('Núm. Eventos');
```

```python
df.groupby(['Ano', 'Tipología']).size().unstack().plot(
    kind='bar',
    stacked=True,
    color=['steelblue', 'orange']
)
```
